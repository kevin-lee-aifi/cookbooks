# 1. Tar compress entire IDE

```bash
tar --use-compress-program="pigz -3 -p 60" -cf ide_backup.tar.gz /home/jupyter
```

# 2. Transfer tar file from old ide to gbucket

```bash
gcloud auth login

gsutil -o GSUtil:parallel_composite_upload_threshold=2G -o GSUtil:parallel_thread_count=48 -o GSUtil:parallel_process_count=16 -m cp -r /home/jupyter/ide_backup.tar  gs://uf-219b-kevinle-815275457-219b3400-2a33-483a-a457-d5272e41d7db
```

# 3. Transfer tar file from gbucket to new ide

```bash
gcloud auth login

gcloud storage cp gs://uf-219b-kevinle-815275457-219b3400-2a33-483a-a457-d5272e41d7db/ide_backup.tar ~/ide_backup.tar
```

# 4. Uncompress tar xz file

```bash
unpigz -p 60 -c ide_backup.tar.gz | tar xf -
```
