# Globus Installation and Startup

## Downloading globus

```bash
wget https://downloads.globus.org/globus-connect-personal/v3/linux/stable/globusconnectpersonal-latest.tgz
```

## Unzip globus tar file

```bash
tar -xvf globusconnectpersonal-latest.tgz
```

## Set up globus endpoint
```bash
# Set current directory as jupyter w/o root (globus commands don't work in the root directory)
su jupyter

/home/jupyter/globusconnectpersonal-3.2.3/globusconnectpersonal -setup

chmod 777 /home/jupyter/nohup.out
```

## Start newly created globus endpoint

```bash
nohup /home/jupyter/globusconnectpersonal-3.2.3/globusconnectpersonal -start
```

# Globus CLI

## Install globus CLI
```bash
pip install globus-cli
```

## Login to globus without local server
```bash
globus login --no-local-server
```

## Transfering files from NWGC collection to IDE instance
```bash
globus transfer \
"9944243a-c496-4c3b-a005-b06de2a0ac9b:/sequencing_only/allen_institute/aifi/FC00549_10X_22CCGHLT4_20240604.tar" \ # NWGC endpoint UUID then path to file
"f8d90f8c-42cc-11ef-9640-453c3ae125a5:~/FC00549_10X_22CCGHLT4_20240604.tar" # User endpoint UUID then path to file
```

## Showing files in NWGC collection
```bash
globus ls "9944243a-c496-4c3b-a005-b06de2a0ac9b:/sequencing_only/allen_institute/aifi/"
```

## Show transfer task

```bash
globus task show 3059368e-9bb1-11ef-9b63-8fdec6a61c4f
```
