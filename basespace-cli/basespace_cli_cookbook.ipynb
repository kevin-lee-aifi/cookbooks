### BaseSpace CLI Installation

> **NOTE**: You can skip this if you follow the instructions on the basespace_cli_installation.sh

```bash
# Create new directory to store basespace cli
mkdir -p /home/jupyter/basespace

# Download basespace cli
wget "https://launch.basespace.illumina.com/CLI/latest/amd64-linux/bs" -O /home/jupyter/basespace/bs

# Enables executable permissions for basespace cli file
chmod u+x /home/jupyter/basespace/bs
```

### BaseSpace CLI Getting Started

```bash
# Authenticate BaseSpace account in terminal
bs auth
 
# Obtain the project ID through CLI
bs list projects # Lists all projects in the authenticated account
 
# Download fastq's from project
bs download project -i 439003024 -o "/home/workspace/EXP-01152" --extension fastq.gz

# Download fastq's from run
bs download run -i 282861642 -o "/home/jupyter/EXP-01152" --extension fastq.gz
```
