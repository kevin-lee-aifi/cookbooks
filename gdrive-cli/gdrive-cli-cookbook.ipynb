# gdrive-cli

gdrive is a command line application for interacting with Google Drive. Installation and set up instructions are based on their <u>[Github README](https://github.com/glotlabs/gdrive?tab=readme-ov-file)</u> and <u>[Google API Credentials document](https://github.com/glotlabs/gdrive/blob/main/docs/create_google_api_credentials.md)</u>.

## Installation and Set Up

### Install gdrive-cli binary for Windows (Local)

1. Download files from the <u>[gdrive github releases](https://github.com/glotlabs/gdrive/releases/download/3.9.1/gdrive_windows-x64.zip)</u>.
2. Extract zipped file.
4. In the terminal, paste the path to the .exe and add `account add`; run the account add command.
```bash
"C:\Users\kevin.lee\Downloads\gdrive_windows-x64\gdrive.exe" account add
```

5. Change the current directory to a directory where you can find the account export file (ex. Downloads, Desktop, etc.).
6. Paste the path to the .exe and add `account export <ACCOUNT_EMAIL>`; run the account export command.
```bash
"C:\Users\kevin.lee\Downloads\gdrive_windows-x64\gdrive.exe" account export kevin.lee@alleninstitute.org
```

7. Move the account export file into the IDE and run the account import command.
> **Note**: Gdrive needs to be installed on the IDE.
> ```bash
> gdrive account import <ARCHIVE_PATH>
> ```

### Install gdrive-cli binary for Linux (Remote IDE)

1. Execute the gdrive-cli-remote-install.sh script using `source gdrive-cli-remote-install.sh`.
> **Note**: The shell script contains jupyter virtual environment commands and installs gdrive-cli into the virtual environment, but can be commented out if you want to install gdrive-cli to your root environment

## Basic Commands

### `gdrive <COMMAND>`

```
Usage: gdrive <COMMAND>

Commands:
  about        Print information about gdrive
  account      Commands for managing accounts
  drives       Commands for managing drives
  files        Commands for managing files
  permissions  Commands for managing file permissions
  version      Print version information
  help         Print this message or the help of the given subcommand(s)

Options:
  -h, --help  Print help information
```
---

### `gdrive account <COMMAND>`

```
Commands for managing accounts

Usage: gdrive account <COMMAND>

Commands:
  add      Add an account
  list     List all accounts
  current  Print current account
  switch   Switch to a different account
  remove   Remove an account
  export   Export account, this will create a zip file of the account which can be imported
  import   Import account that was created with the export command
  help     Print this message or the help of the given subcommand(s)

Options:
  -h, --help  Print help information
```
---

### `gdrive drives <COMMAND>`

```
Commands for managing drives

Usage: gdrive drives <COMMAND>

Commands:
  list  List drives
  help  Print this message or the help of the given subcommand(s)

Options:
  -h, --help  Print help information
```
---

### `gdrive files <COMMAND>`

```
Commands for managing files

Usage: gdrive files <COMMAND>

Commands:
  info      Print file info
  list      List files
  download  Download file
  upload    Upload file
  update    Update file. This will create a new version of the file. The older versions will typically be kept for 30 days
  delete    Delete file
  mkdir     Create directory
  rename    Rename file/directory
  move      Move file/directory
  copy      Copy file
  import    Import file as a google document/spreadsheet/presentation. Example of file types that can be imported: doc, docx, odt, pdf, html, xls, xlsx, csv, ods, ppt, pptx, odp
  export    Export google document to file
  help      Print this message or the help of the given subcommand(s)

Options:
  -h, --help  Print help information
```
